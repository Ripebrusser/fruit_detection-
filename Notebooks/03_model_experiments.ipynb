# 03 · Model experiments and results

Loads the training history and evaluation artefacts produced by
`python -m SRC.models.train` and `python -m SRC.models.evaluate`.
All numbers shown here are read from files written by the evaluation run — nothing is typed in by hand.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'Notebooks' else Path.cwd()
metrics = json.loads((PROJECT_ROOT / 'Evaluation/metrics/metrics.json').read_text())
print(json.dumps({k: metrics[k] for k in ('model', 'accuracy', 'f1_macro', 'fruit_level_accuracy', 'condition_level_accuracy')}, indent=2))
```
