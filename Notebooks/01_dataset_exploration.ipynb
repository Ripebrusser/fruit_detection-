# 01 · Dataset exploration

Reads the artefacts produced by `python -m SRC.data.inspect_dataset` and reviews the
dataset composition: class balance, image sizes, duplicates and content type.

Run from the project root with the project virtual environment selected as the kernel.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'Notebooks' else Path.cwd()
summary = json.loads((PROJECT_ROOT / 'Dataset/interim/inspection_summary.json').read_text())
print('images:', summary['n_files_total'], '| readable:', summary['n_readable'], '| formats:', summary['extensions'])
print('size range:', summary['image_size_stats'])
```

```python
counts = pd.Series(summary['per_class_counts']).sort_index()
counts.to_frame('images')
```

## Duplicates
Exact duplicates (identical SHA-256) and near duplicates (pHash) are used to keep the train/validation/test split leak-free.

```python
dups = summary['duplicates']
print('exact duplicate groups:', dups['n_exact_groups'], '| redundant files:', dups['n_exact_duplicate_files'])
print('near duplicate groups (pHash<=8):', dups['n_near_groups'], '| spanning classes:', dups['n_near_groups_cross_class'])
for group in dups['exact_groups'][:3]:
    print(group)
```
